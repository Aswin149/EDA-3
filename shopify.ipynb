```python
df["date"]=pd.to_datetime(df["date"],utc=True)
```

```python
df['date'].dtypes
```

```python
print(df.isnull().sum())
```

```python
df=df.sort_values("date")
```

```python
df["Date_delta"]=df["close"]-df["open"]
```

```python
df["Daily_delta"]=((df['close']-df['open'])/df['open'])*100
```

```python
df["Daily_returns"]=((df['close']-df['open'])/df['open'])*100
```

```python
print(df[
    ["date","open","close","Daily_delta","Daily_returns"]
].head())
```

```python
average_volume=df["volume"].mean()
maximum_volume=df["volume"].max()
minimum_volume=df["volume"].min()
```

```python
print("Average Volume:",average_volume)
print("Maximum Volume:",maximum_volume)
print("Minimum Volume:",minimum_volume)
```

```python
threshold=2*average_volume
```

```python
anomalous_days=df[df["volume"]>threshold]
```

```python
print("Anomalous Days:")
print(anomalous_days[["date","volume"]])
```

```python
print("Mean Return:",df["Daily_returns"].mean())
print("Median Return:",df["Daily_returns"].var())
print("Standard Deviation:",df["Daily_returns"].std())
```
